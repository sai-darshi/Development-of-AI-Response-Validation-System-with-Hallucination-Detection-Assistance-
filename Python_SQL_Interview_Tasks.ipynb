Question 2: Reverse a String

Prompt: Write a Python program to reverse a given string.

Concept: String slicing

```python
text = "Python"
reversed_text = text[::-1]

print("Original string:", text)
print("Reversed string:", reversed_text)
```

Question 3: Find the Largest Number in a List

Prompt: Write a Python program to find the largest number in a list of numbers.

Concept: Lists and the "max()" function

```python
numbers = [10, 25, 8, 40, 15]

largest = max(numbers)

print("Largest number:", largest)
```

Question 4: Check Palindrome

Prompt: Write a Python program to check whether a given string is a palindrome.

Concept: String slicing and if-else statements

```python
word = "madam"

if word == word[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

Question 5: Count Vowels in a String

Prompt: Write a Python program to count the number of vowels in a given string.

Concept: Strings, loops, conditions, and the "sum()" function

```python
text = "education"
vowels = "aeiou"

count = sum(1 for ch in text.lower() if ch in vowels)

print("Number of vowels:", count)
```

SQL Interview Questions

SQL (Structured Query Language) is used to store, retrieve, and manage data in databases.

The following questions demonstrate important SQL queries commonly asked in interviews.

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE Employee (
    ID INTEGER,
    Name TEXT,
    Department TEXT,
    Salary INTEGER
)
""")

cursor.executemany("""
INSERT INTO Employee VALUES (?, ?, ?, ?)
""", [
    (1, "Ravi", "IT", 25000),
    (2, "Priya", "HR", 40000),
    (3, "Arun", "IT", 35000),
    (4, "Sita", "Sales", 30000),
    (5, "Kiran", "IT", 50000)
])

conn.commit()

print("Employee table created successfully.")
```

SQL Question 1: Display All Employees

Prompt: Write an SQL query to display all records from the Employee table.

Concept: SELECT statement

```python
query = "SELECT * FROM Employee"

pd.read_sql_query(query, conn)
```

SQL Question 2: Filter Employees by Salary

Prompt: Write an SQL query to display the names and salaries of employees whose salary is greater than 30,000.

Concept: SELECT and WHERE statements

```python
query = """
SELECT Name, Salary
FROM Employee
WHERE Salary > 30000
"""

pd.read_sql_query(query, conn)
```

SQL Question 3: Find the Highest Salary

Prompt: Write an SQL query to find the employee who earns the highest salary.

Concept: ORDER BY and LIMIT

```python
query = """
SELECT Name, Salary
FROM Employee
ORDER BY Salary DESC
LIMIT 1
"""

pd.read_sql_query(query, conn)
```

SQL Question 4: Count Employees in Each Department

Prompt: Write an SQL query to count the number of employees in each department.

Concept: GROUP BY and COUNT()

```python
query = """
SELECT Department, COUNT(*) AS Employee_Count
FROM Employee
GROUP BY Department
ORDER BY Department
"""

pd.read_sql_query(query, conn)
```

SQL Question 5: Find the Second-Highest Salary

Prompt: Write an SQL query to find the second-highest distinct salary from the Employee table.

Concept: Subquery and MAX() function

```python
query = """
SELECT MAX(Salary) AS Second_Highest_Salary
FROM Employee
WHERE Salary < (SELECT MAX(Salary) FROM Employee)
"""

pd.read_sql_query(query, conn)
```
